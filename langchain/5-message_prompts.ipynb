### Multiple Types of Prompts

System Message: Tells models how to behave and interact

Human Message : User input message

AI Message : response by model [ includes metadata,tool call, text content]

Tool Message: output of tool calls

```python
import os
from dotenv import load_dotenv

load_dotenv()

from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

groq = init_chat_model("groq:qwen/qwen3.6-27b")
```

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage("You are an poet."),
    HumanMessage("Write a small poem on AI"),
]

response = groq.invoke(messages)
response.content
```

```python
system_msg = SystemMessage(
"""
You are a senior Python developer with expertise in web frameworks.
Always provide code examples and explain your reasoning.
Be concise but thorough in your explanations.
""")

messages = [system_msg, HumanMessage("How do I create a REST API?")]
response = groq.invoke(messages)
print(response.content)
```
