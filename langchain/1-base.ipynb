```python
import os
from dotenv import load_dotenv

load_dotenv()

# os.environ["OPENROUTER_API_KEY"] = os.getenv("OPENROUTER_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
```

```python
from langchain.agents import create_agent


def get_weather(city: str) -> str:
    """Get weather for a city"""

    return f"The weather for the city of {city} is this."


agent = create_agent(
    model="google_genai:gemini-flash-latest",
    tools=[get_weather],
    system_prompt="You are a helpful assistant",
)
# agent : gives the architecture of the agent

# Both works
# agent_response = agent({"message" : "what's the weather in NYC?"})
agent_response = agent.invoke(
    {"messages": [{"role": "user", "content": "what is the weather in NYC?"}]}
)

# agent_response["message"] -> gives whole api output.
# agent_response["messages"][-1].content  --> gives you the agent's response
```
