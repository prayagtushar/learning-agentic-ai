```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv()

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")

gemini = init_chat_model("google_genai:gemini-3.5-flash")
```

### Streaming
<!-- Print chunk by chunk as the model spits out -->
<!-- flush=True forces each chunk.text to appear as soon as it is received. -->

```python
for chunk in gemini.stream("Write a 150 word paragraph on rise of AI"):
    print(chunk.text, end=" ", flush=True)
```

### Batch
<!-- Collecting bunch of requests and sending them together to the model. -->
<!-- Helps improve performance and reduce costs since it's being done parallely. -->

```python
requests = [
    "Why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "What is quantum computing?",
]

responses = gemini.batch(requests)

for resp in responses:
    print(resp)
```

### Concurrency
We can decide how many requests should reach the model at a single time.

```python
requests = [
    "Why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "What is quantum computing?",
]

responses = gemini.batch(
    requests, config={"max_concurrency": 2}  # Only two req will be send a time.
)

for resp in responses:
    print(resp)
```
